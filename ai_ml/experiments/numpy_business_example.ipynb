# Advance operation with Business examples

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Mini Sample Sales Data Set
# Data structure: [restaurant_id, 2021, 2022, 2023, 2024]
data = np.array([
    [1, 150000, 180000, 220000, 250000],  # Paradise Biryani
    [2, 120000, 140000, 160000, 190000],  # Beijing Bites
    [3, 200000, 230000, 260000, 300000],  # Pizza Hub
    [4, 180000, 210000, 240000, 270000],  # Burger Point
    [5, 160000, 185000, 205000, 230000]   # Chai Point
])
```

```python
# Shape of the given data (Most common operation)
print("Shape of the given sales data:", data.shape)

# Printing the sales data of first 3 resturants
print("\nThe sales data of first 3 resturants:\n", data[:3]) # this can also be written `data[0:3]`
```

```python
# Removed all the resturant_id
sales_data = data[:, 1:]

# Total sales per year
total = np.sum(sales_data, axis=0) # Sum only values
print("Total sales per year:", total)

# Minimum sales per restaurant
min_sales = np.min(sales_data, axis=1)
print("\nMinimum sales per restaurant:", min_sales)

# Minimum sales per year
min_sales_year = np.min(sales_data, axis=0)
print("\nMinimum sales per year:", min_sales_year)

# Maximum sales per restaurant
max_sales = np.max(sales_data, axis=1)
print("\nMaximum sales per restaurant:", max_sales)

# Maximum sales per year
max_sales_year = np.max(sales_data, axis=0)
print("\nMaximum sales per year:", max_sales_year)

# Average sales per restautant
avg_sales = np.mean(sales_data, axis=1)
print("\nAverage sales per restaurant:", avg_sales)

# Average sales per year
avg_sales_year = np.mean(sales_data, axis=0)
print("\nAverage sales per year:", avg_sales_year)
```

<span style="color:red;">***Note***</span>

Both `np.mean` and `np.average` from NumPy compute averages, but they differ in flexibility:

- `np.mean` is simpler and slightly faster and does not support weights

- `np.average` is only useful when you need weights


> A **`weight`** is a value that tells how important or influential each data point should be when calculating an average.


**Simple idea:**

- In a normal average (like `np.mean`), all values are treated equally.

- In a weighted average, some values count more than others.

```python
# Cumulative Sum
cumsum = np.cumsum(sales_data, axis=1)
print("Cumulative sum:\n", cumsum)
```

```python
# Plotting cumsum using matplotlib
plt.figure(figsize=(8,5))
plt.plot(np.mean(cumsum, axis=0))
plt.title("Average Cumulative Sales across all Restaurants")
plt.xlabel("Years")
plt.ylabel("Sales")
plt.grid(True)
plt.show()
```

```python
# Plotting Graph
plt.figure(figsize=(8,5))
plt.plot(np.max(sales_data, axis=1))
plt.title("Maximum sales per year")
plt.xlabel("Year")
plt.ylabel("Sales")
plt.grid(True)
plt.show()
```

```python
# Plotting Graph
plt.figure(figsize=(8,5))
plt.plot(np.min(sales_data, axis=0))
plt.title("Minimum sales per restaurant")
plt.xlabel("Year")
plt.ylabel("Sales")
plt.grid(True)
plt.show()
```

## Broadcast

Broadcasting in NumPy is a mechanism that allows arrays of different shapes to be used together in arithmetic operations.

```python
monthly_avg = sales_data / 12
print("Monthly average:\n", monthly_avg)
```

## Truncate
In NumPy, `truncate` usually refers to removing the decimal part of numbers—essentially chopping off everything after the decimal point without rounding.

```python
# Truncate
monthly_avg = np.trunc(sales_data / 12)
print("Truncated Monthly average:\n", monthly_avg)
```
