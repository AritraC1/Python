## Analysis and Visualisation

```python
# Type of pokemon - Normal/Legendary/Mythical with pie and percentage

# Normal (Need to make this category - is_legendary: false & is_Mythical: false)
normal = df[(df['is_legendary'] == False) & (df['is_mythical'] == False)]

# Legendary
legendary = df[df['is_legendary'] == True]

# Mythical
mythical = df[df['is_mythical'] == True]

counts = [len(normal), len(legendary), len(mythical)]
labels = ['Normal', 'Legendary', 'Mythical']

# Pie chart with percentage
plt.pie(counts, labels=labels, autopct='%1.2f%%')
plt.title('Pokemon Type Distribution')

# Caption
plt.figtext(
    0.5, -0.05,
    "Figure 1 — Pokemon Type Distribution (Normal, Legendary, Mythical)",
    ha="center",
    fontsize=10
)

plt.show()
```

```python
# Finding: Dataset is heavily imbalanced
# A model predicting "Normal" every time = ~88% accurate but useless
# This is called the accuracy paradox (for ML)
```

```python
# type with the strongest Pokémon on average
type_strength = df.groupby('type_1')['base_stat_total'].mean().sort_values()

type_strength.plot(kind='barh', figsize=(10, 7), color='steelblue')

plt.title('Average Base Stat Total by Type')
plt.xlabel('Avg Base Stat Total')
plt.tight_layout()

plt.show()
```

```python
# Do Pokemon get stronger each generation?

gen_strength = df.groupby('generation')['base_stat_total'].mean()

gen_strength.plot(kind='bar', figsize=(9, 5), color='steelblue')

plt.title('Average Base Stat Total by Generation')
plt.xlabel('Generation')
plt.ylabel('Avg Base Stat Total')
plt.xticks(rotation=0)
plt.tight_layout()

plt.show()
```

```python
color_counts = df['color'].value_counts()

bar_colors = {
    'red': 'red', 'blue': 'blue', 'yellow': 'gold',
    'green': 'green', 'black': 'black', 'brown': 'brown',
    'purple': 'purple', 'gray': 'gray', 'white': 'lightgray',
    'pink': 'pink'
}
colors = [bar_colors.get(c, 'steelblue') for c in color_counts.index]

color_counts.plot(kind='bar', color=colors, figsize=(10, 5), edgecolor='white')

plt.title('Most Common Pokemon Colors')
plt.xlabel('Color')
plt.ylabel('Count')
plt.xticks(rotation=45)
plt.tight_layout()

plt.show()
```

```python
# Single-type vs dual-type — who is stronger

# is_dual_type already exists in dataset
single = df[df['is_dual_type'] == False]['base_stat_total']
dual   = df[df['is_dual_type'] == True]['base_stat_total']

# Box plot shows distribution, median, and outliers
plt.figure(figsize=(7, 5))
plt.boxplot([single, dual], labels=['Single Type', 'Dual Type'])
plt.title('Single Type vs Dual Type — Strength Distribution')
plt.ylabel('Base Stat Total')
plt.tight_layout()

plt.show()
```

```python
# Stat tier breakdown

tier_order = [
    'Weak (<300)',
    'Below Average (300-399)',
    'Average (400-499)',
    'Strong (500-599)',
    'Legendary/Pseudo (600+)'
]

tier_counts = df['stat_tier'].value_counts().reindex(tier_order)

tier_counts.plot(kind='bar', figsize=(10, 5), color='steelblue', edgecolor='white')

plt.title('Pokemon Count by Stat Tier')
plt.xlabel('Stat Tier')
plt.ylabel('Count')
plt.xticks(rotation=20, ha='right')
plt.tight_layout()

plt.show()
```
